```python
import CoolProp.CoolProp as CP
from scipy.optimize import fsolve
import numpy as np
from scipy.optimize import least_squares
import matplotlib.pyplot as plt
import math
# Define the fluid
fluid = 'Argon'

# Example: Get properties at a specific temperature and pressure
temperature = 2673.15  # Convert to Kelvin
pressure = 1e6  # Pressure in Pa (25 MPa)

# Get density (kg/m^3)
density = CP.PropsSI('D', 'T', temperature, 'P', pressure, fluid)

# Get specific heat capacity (J/kg.K)
cp = CP.PropsSI('C', 'T', temperature, 'P', pressure, fluid)

# Get dynamic viscosity (Pa.s)
viscosity = CP.PropsSI('V', 'T', temperature, 'P', pressure, fluid)

# Get thermal conductivity (W/m.K)
thermal_conductivity = CP.PropsSI('L', 'T', temperature, 'P', pressure, fluid)

# Get enthalpy (J/kg)
enthalpy = CP.PropsSI('H', 'T', temperature, 'P', pressure, fluid)

print(f"Density: {density} kg/m^3")
print(f"Specific Heat Capacity: {cp} J/kg.K")
print(f"Viscosity: {viscosity} Pa.s")
print(f"Thermal Conductivity: {thermal_conductivity} W/m.K")
print(f"Enthalpy: {enthalpy} J/kg")
```

```python
m_gr = 18 #kg
cp_gr = 2000 #J/kgK
V_gas_init = 200 #m^3
T_gr_init = 2073 #K
T_w_init = 293
T_std = 293
P_init = 1e5 #Pa
P_atm = 1e5
A_film = 0.413 #m^2
eps_gr = 0.8
sb = 5.67e-8
R_w = 8.304 / 0.018 #J/kgK
R_ar = 8.304 / .03995
P_relief = 1.002e5 #MPa
P_burst = 1.005e5 #MPa
P_burst2 = 1.005e5
C_d_relief = 0.5
C_v_solenoid = 0
C_d_burst = 0.9
k_w = 1.324
k_ar = 1.66
d_relief = 6 #in
d_ex = 0.0508#m
L_ex = 8
N_elbows = 10
K_elbow = 0.3
K_tee = 1.0
d_tee = 0.0414
N_reliefs = 1
A_relief = N_reliefs * np.pi * (d_relief * 0.0254) ** 2 / 4
d_burst = 3.63 #in
d_burst2 = 3.5
N_burst = 1
N_burst2 = 1
#vdot_max_burst = 0.5 * N_burst
#vdot_max_burst = 5 * N_burst
A_burst = N_burst * np.pi * (d_burst * 0.0254) ** 2 / 4
A_burst2 = N_burst2 * np.pi * (d_burst2 * 0.0254) ** 2 / 4
is_temp_equilibrium = True

t_end = 240
t_log = t_end * 0.1
n_steps_log = 301
n_steps_lin = 5*n_steps_log
dt_lin = (t_end - t_log) / n_steps_lin
time_steps = np.concatenate(([0],np.logspace(.01,np.log10(t_log),n_steps_log) - 1,np.linspace(t_log + dt_lin,t_end,n_steps_lin)))
m_w_list_gpm = [0.05, 0.1,0.2,0.3,1,3,5]
#m_w_list_gpm = [.05,0.1,0.2,0.3, 1,2,5,10,30]
P = np.empty((len(time_steps)+1, len(m_w_list_gpm)))
T_gr = np.empty((len(time_steps)+1, len(m_w_list_gpm)))
T_mix = np.empty((len(time_steps)+1, len(m_w_list_gpm)))
V_gas=V_gas_init
m_ar = V_gas * CP.PropsSI('D', 'T', T_gr_init, 'P', P_init, 'Argon') #intial Argon mass

def CircPipeFF(Re, D_h, rough):
    #return circular pipe friction factor (turbulent: Zigrang and Sylvester)
    if (Re < 2300):
        return 64 / Re
    else:
        return (-2 * np.log10(2 * rough / (7.54 * D_h) - 5.02 / Re * np.log10(2*rough / (7.54 * D_h) + 13 / Re))) ** (-2)
for j, m_rate_gpm in enumerate(m_w_list_gpm):
    m_rate = 3.9 / 60 * m_rate_gpm
    P[0,j] = P_init
    T_gr[0,j] = T_gr_init
    mdot_relief = 0
    is_burst = False
    is_burst2 = False
    m_w=0 #mass of liquid water in vessel
    tot_boil = 0 #mass of steam generated
    for i, time in enumerate(time_steps):
        if i < len(time_steps)-1:
            dt = time_steps[i+1] - time
        else:
            dt = time - time_steps[i-1]
        if V_gas == 0.1:
            m_rate = 0
        m_w=max(0,m_w) + m_rate*dt
        m_tot = m_ar + tot_boil # total gas exhausted and in vessel
        f_ar = m_ar / m_tot #gas fraction Argon
        f_w = tot_boil / m_tot #gas fraction steam
        k_mix = k_w * f_w + k_ar * f_ar
        R_mix = R_w * f_w + R_ar * f_ar
        if P[i,j] > P_burst:
            is_burst = True
        if P[i,j] > P_burst2:
            is_burst2 = True
        #Calculate steam generation rate
        T_sat = CP.PropsSI('T', 'Q', 1, 'P', P[i,j], 'Water')
        if is_temp_equilibrium:
            T_steam = T_gr[i,j]
        else:
            T_steam = T_sat + 1
        dh_water = CP.PropsSI('H', 'T', T_steam, 'P', P[i,j], 'Water') - CP.PropsSI('H', 'T', T_w_init, 'P', P[i,j], 'Water')
        k_v = CP.PropsSI('L', 'T', T_sat+1, 'P', P[i,j], 'Water')
        rho_v = CP.PropsSI('D', 'T', T_sat+1, 'P', P[i,j], 'water')
        rho_l = CP.PropsSI('D', 'T', T_sat-1, 'P', P[i,j], 'water')
        nu_l = CP.PropsSI('V', 'T', T_sat-1, 'P', P[i,j], 'water')
        st_v = CP.PropsSI('I', 'T', T_sat, 'Q', 0.99, 'water')
        h_fg = CP.PropsSI('H', 'T', T_sat+1, 'P', P[i,j], 'Water') - CP.PropsSI('H', 'T', T_sat - 1, 'P', P[i,j], 'Water')
        dh_vap = (CP.PropsSI('H', 'T', T_steam, 'P', P[i,j], 'Water') - CP.PropsSI('H', 'T', T_sat + 1, 'P', P[i,j], 'Water'))
        h_film_conv = 0.425 * (k_v ** 3 * rho_v * 9.8 * (rho_l - rho_v) * (h_fg + 0.4 * dh_vap) / (nu_l * rho_l * (T_steam - T_sat) * 
                                                                                                   (st_v / (9.8 * (rho_l - rho_v))) ** 0.5)) ** 0.25 
        h_film_rad = eps_gr * sb * (T_gr[i,j] ** 4 - T_sat ** 4) / (T_gr[i,j] - T_sat)
        h_film = 100
        for _ in range(100):
            h_film_last = h_film
            h_film = (h_film_conv ** (4/3) + h_film_rad * h_film ** (1/3)) ** 0.75
            if abs(h_film - h_film_last) < 1:
                break
        q_film = h_film * (T_gr[i,j] - T_sat)
        rate_boil = min(q_film * A_film / dh_water, max(m_w, 0) / dt) #steam gen rate in kg/s
        if T_gr[i,j] > T_sat + 100: #at this point nucleate boiling occurs, cooling rapidly but such low temperature that it doesn't drive results
            T_gr[i + 1,j] = T_gr[i,j] - rate_boil * dt * dh_water / (m_gr * cp_gr)
        else:
            T_gr[i + 1,j] = T_gr[i,j]
            rate_boil = 0
        T_mix[i,j] = T_steam * f_w + T_gr[i,j] * f_ar
        #update amount of liquid water in vessel
        m_w -= rate_boil * dt 
        #update gas volume
        V_gas = max(0.1,V_gas_init - max(0,m_w) / CP.PropsSI('D', 'T', T_w_init, 'P', P[i,j], 'water'))
        dPdt = rate_boil * R_mix * T_mix[i,j] / V_gas
        tot_boil += rate_boil * dt
        #density of gas mixture
        rho_mix = CP.PropsSI('D', 'T', T_steam, 'P', P[i,j], 'water') * f_w + CP.PropsSI('D', 'T', T_gr[i,j], 'P', P[i,j], 'Argon') * f_ar
        rho_mix_std = CP.PropsSI('D', 'T', T_sat + 1, 'P', P_atm, 'water') * f_w + CP.PropsSI('D', 'T', T_std, 'P', P_atm, 'Argon') * f_ar
        rho_std = CP.PropsSI('D', 'T', T_std, 'P', P_atm, 'Air')
        SG = rho_mix_std / rho_std #specific gravity
        rho_mix_ex = CP.PropsSI('D', 'T', T_steam, 'P', P_atm, 'water') * f_w + CP.PropsSI('D', 'T', T_gr[i,j], 'P', P_atm, 'Argon') * f_ar
        mu_mix_ex = (CP.PropsSI('V', 'T', T_steam, 'P', P_atm, 'water') * f_w + CP.PropsSI('V', 'T', T_gr[i,j], 'P', P_atm, 'Argon') * f_ar) * rho_mix_ex
        P_crit = P_atm / (2/(k_mix + 1)) ** (k_mix/(k_mix - 1))
        if P[i, j] > P_relief:
            if P[i,j] > P_crit: #choked
                mdot_relief = C_d_relief * A_relief * (2 * rho_mix * (P_crit - P_atm)) ** 0.5
                vdot_solenoid_SCFH = 962 * C_v_solenoid * (((P_crit/6870) ** 2 - (P_atm/6870) ** 2) / (SG * 1.8 * T_gr[i,j])) ** 0.5
                mdot_solenoid = 7.867e-6 * vdot_solenoid_SCFH * rho_mix_std
            else:
                mdot_relief = C_d_relief * A_relief * (2 * rho_mix * (P[i,j] - P_atm)) ** 0.5
                vdot_solenoid_SCFH = 962 * C_v_solenoid * (((P[i,j]/6870) ** 2 - (P_atm/6870) ** 2) / (SG * 1.8 * T_gr[i,j])) ** 0.5
                mdot_solenoid = 7.867e-6 * vdot_solenoid_SCFH * rho_mix_std
            vdot_relief_SCFH = mdot_relief / rho_mix_std * 3600 / .3048**3
            dPdt -= (mdot_relief + mdot_solenoid) * R_mix * T_mix[i,j] / V_gas
            #print(P[i,j],P_atm,SG,f_ar,f_w,T_gr[i,j],C_v_solenoid, vdot_solenoid_SCFH/60,mdot_solenoid)
            #print("relief",P[i,j],dPdt, rate_boil,mdot_relief,mdot_solenoid,vdot_relief_SCFH,vdot_solenoid_SCFH,R_mix, rho_mix, rho_mix_ex)
        if is_burst:
            P_ex = (P_atm + P[i,j]) / 2
            for _ in range(100):
                P_ex_last = P_ex
                mdot_ex = 0.1
                dP_tee = 0.0
                P_crit = P_ex / (2/(k_mix + 1)) ** (k_mix/(k_mix - 1))
                if P[i,j] - dP_tee > P_crit: #choked
                    R_hyd_burst = (P_crit - P_ex) / (2 * (C_d_burst * A_burst) ** 2 * (P[i,j] - dP_tee - P_ex))
                elif P[i,j] - dP_tee > P_ex: #unchoked
                    R_hyd_burst = 1 / (2 * (C_d_burst * A_burst) ** 2)
                    
                if P[i,j] > P_ex:
                    Re_ex = 4 * mdot_ex / (np.pi * mu_mix_ex * d_ex)
                    ff = CircPipeFF(Re_ex, d_ex, 1e-5)
                    R_hyd_pipe = 8 * ff * L_ex / (np.pi ** 2 * d_ex ** 5) + N_elbows * K_elbow / (np.pi ** 2 * d_ex ** 4)
                    R_hyd_tee = 8 * K_tee / (np.pi ** 2 * d_tee ** 4)
                    R_hyd_tot = R_hyd_burst + R_hyd_pipe + R_hyd_tee
                    mdot_ex = (rho_mix_ex * (P[i,j] - P_atm) / R_hyd_tot) ** 0.5
                    dP_pipe = mdot_ex ** 2 * R_hyd_pipe / rho_mix_ex
                    dP_tee = mdot_ex ** 2 * R_hyd_tee / rho_mix_ex
                    P_ex = P_atm + dP_pipe
                else:
                    mdot_ex = 0
            #print("burst",P[i,j], P_ex,dP_tee, dP_pipe, Re_ex,mdot_ex)
            dPdt -= mdot_ex * R_mix * T_mix[i,j] / V_gas
        if is_burst2:
            P_crit = P_atm / (2/(k_mix + 1)) ** (k_mix/(k_mix - 1))
            if P[i,j] > P_crit: #choked
                R_hyd_burst2 = (P_crit - P_atm) / (2 * (C_d_burst * A_burst2) ** 2 * (P[i,j] - P_atm))
                mdot_ex = (rho_mix_ex * (P[i,j] - P_atm) / R_hyd_burst2) ** 0.5
            elif P[i,j] > P_atm: #unchoked
                R_hyd_burst2 = 1 / (2 * (C_d_burst * A_burst2) ** 2)
                mdot_ex = (rho_mix_ex * (P[i,j] - P_atm) / R_hyd_burst2) ** 0.5
            else:
                mdot_ex = 0
            #print("burst2",P[i,j],mdot_ex)
            dPdt -= mdot_ex * R_mix * T_mix[i,j] / V_gas
        P[i + 1,j] = P[i,j] + dPdt * dt
P_gauge = (P - P_atm)*14.7/1e5
# Plotting the results
plt.figure(figsize=(10, 6))
for j, m_w in enumerate(m_w_list_gpm):
    plt.semilogx(time_steps, P_gauge[:-1, j], label=f'{m_w:.4f} GPM')

plt.xlabel('Time (s)')
plt.ylabel('Pressure (Psi)')
plt.title('Pressure Over Time for Different Water Flowrates')
plt.legend(loc='upper left')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
for j, m_w in enumerate(m_w_list_gpm):
    plt.semilogx(time_steps, T_gr[:-1, j]-273, label=f'{m_w:.4f} GPM')

plt.xlabel('Time (s)')
plt.ylabel('Graphite Temperature, (C)')
plt.title('Graphite Temp Over Time for Different Water Flowrates')
plt.legend(loc='lower left')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
for j, m_w in enumerate(m_w_list_gpm):
    plt.semilogx(time_steps, T_mix[:-1, j]-273, label=f'{m_w:.4f} GPM')

plt.xlabel('Time (s)')
plt.ylabel('Gas Temperature, (C)')
plt.title('Gas Temp Over Time for Different Water Flowrates')
plt.legend(loc='lower left')
plt.grid(True)
plt.show()
```
