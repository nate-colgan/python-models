```python
import numpy as np
import CoolProp.CoolProp as CP
import thermo as thermo
```

```python
mdot_Sn = 1 #kg/s
T_Sn = 800 # K
P_Sn = 2e5 # Pa

rho_Sn = CP.PropsSI('D','T', T_Sn, 'P', P_Sn, 'Tin')
```

```python
def Sn_density(temp):
    #return liquid Tin density
    return 6982.2 - 0.75643 * (temp - 505) + 1.074e-4 * (temp - 505) ** 2 - 2.194e-8 * (temp - 505) ** 3

def Sn_tcx(temp):
    #return liquid Tin thermal conductivity
    return 13.9 * 0.02868 * temp

def Sn_mu(temp):
    #return liquid Tin viscosity
    return 3.908e-4 * np.exp(790.7 / temp)

```
