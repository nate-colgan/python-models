```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.constants import sigma

# Parameters
L_tot = 22*.0254           # Total fin length (m)
D = 0.0127              # Fin diameter (m)
k = 30              # Thermal conductivity (W/m/K)
q_root_max = -5e5          # Heat flux at root (W/m^2)
T_tip_C = 2200        # Tip temperature (°C)
T_tip = T_tip_C + 273.15
epsilon_emitter = 0.8
epsilon_fin = 0.8
epsilon_ins = 0.0
epsilon_chamber = 0.3
T_emitter_C = 2200
T_emitter = T_emitter_C + 273.15
T_chamber_C = 300
T_chamber = T_chamber_C + 273.15
T_ins_C = 1100
T_ins = T_ins_C + 273.15
L_ins = 0.1
L_rad = 0.25          # Length from tip with radiative heating (m)
N = 201               # Number of nodes
rho_gr = 1750
cp_gr = 2100

T_root_min_C = 40   # Minimum allowed root temp (°C)
T_root_min = T_root_min_C + 273.15

dx = L_tot / (N-1)
dt=0.1
A = np.pi * (D/2)**2
P = np.pi * D
x = np.linspace(0, L_tot, N)

illuminated = x > (L_tot - L_rad)
#insulation = (L_tot - L_rad - L_ins) < x < (L_tot - L_rad)
unilluminated = x < (L_tot - L_rad - L_ins)

#insulation = !(illuminated or unilluminated)
#print(insulation)

# Initial guess: linear profile
T = np.linspace(T_root_min, T_tip, N)

max_iter = 100000
tol = 1e-3

for it in range(max_iter):
    T_old = T.copy()
    q_rad = np.zeros(N)
    q_rad_tot = np.zeros(N)
    q_cond = np.zeros(N)
    q_fin = np.zeros(N)
    q_fin_check = np.zeros(N)
    q_rad = epsilon_ins * epsilon_fin * sigma * (T_ins**4 - T**4) * P * dx
    q_rad[illuminated] = epsilon_emitter * epsilon_fin * sigma * (T_emitter**4 - T[illuminated]**4) * P * dx
    q_rad[unilluminated] = epsilon_chamber * epsilon_fin * sigma * (T_chamber**4 - T[unilluminated]**4) * P * dx

    T_new = T.copy()
    for i in range(0, N):
        #q_rad_tot[i] = np.sum(q_rad[i:N]) #total heat into rod at x>dx*i
        if i == 0:
            q_cond[i] = (T[0] - T[1]) * k * A / dx
        elif i == N-1:
            q_cond[i] = (T[i] - T[i-1]) * k * A / dx
        else:
            q_cond[i] = (2 * T[i] - T[i+1] - T[i-1]) * k * A / dx
        #q_fin[i] = q_rad_tot[i] - q_cond[i]
        q_fin[i] = q_rad[i] - q_cond[i]
    #print(q_rad)
    #print(q_fin)
    for i in range(1, N):
        # Explicit update (energy balance at node)
        #T_new[i] = (T[i-1] + T[i+1] + (q_rad[i] / (k*A)) * dx**2) / 2
        T_new[i] = min(T_tip,max(T_root_min,T_new[i-1],T[i] + q_fin[i] / (A*dx*rho_gr*cp_gr) * dt))
        #T_new[N-i-1] = T_new[N-i] - q_fin[N-i-1] * dx / (A*k)
        #q_fin_check[N-i-1] = (T_new[N-i] - T_new[N-i-1]) * k * A / dx

    #print(T_new[0])

    T_new[-1] = T_tip

    #print(T_new-273.15)
    #print(np.nanmax(np.abs(T_new - T)))
    if np.nanmax(np.abs(T_new - T)) < tol:
        print(f"Converged after {it+1} iterations")
        break

    T = T_new
    #print(T)

else:
    print("Did not converge!")

# Calculate actual root flux (for reporting)
actual_flux = k * (T[0] - T[1]) / dx / 1e4
q_root = actual_flux *1e4* A

print(f"Root temperature: {T[0]-273.15:.2f} °C")
print(f"Actual root heat flux: {actual_flux:.2f} W/cm^2")
print(np.sum(q_rad), q_root)
print(T)
print(q_rad)
print(q_fin)

plt.plot(x, T-273.15, label='Rod temperature')
plt.axvspan(L_tot-L_rad, L_tot, color='orange', alpha=0.2, label='Illuminated region')
plt.xlabel('Position along rod (m)')
plt.ylabel('Temperature (°C)')
plt.title('Cylindrical Rod with Distributed Radiative Heating')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
